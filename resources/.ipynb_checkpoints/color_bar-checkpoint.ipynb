```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

```python
cmap = plt.cm.viridis
norm = mpl.colors.Normalize(vmin=0, vmax=1)
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

plt.rcParams['figure.dpi'] = 300
plt.rcParams['figure.figsize'] = 1.05, 2
fig, ax = plt.subplots()

# ax.axis('off')

cbar = plt.colorbar(
    sm,
    cax=ax,
    orientation='vertical',
    shrink=0.5,
)

cbar.set_ticks([0, 1])
cbar.set_ticklabels(['low', 'high'])
cbar.ax.tick_params(labelsize=12)
cbar.set_label('expression', rotation=90, labelpad=0, fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.rcParams['figure.dpi'] = 300
plt.rcParams['figure.figsize'] = 3, 0.85
fig, ax = plt.subplots()

cbar = plt.colorbar(
    sm,
    cax=ax,
    orientation='horizontal',
)

cbar.set_ticks([0, 1])
cbar.set_ticklabels(['low', 'high'])
cbar.ax.tick_params(labelsize=12)
cbar.set_label('expression', labelpad=0, fontsize=12)

plt.tight_layout()
plt.show()
```

```python
cmap = plt.cm.bwr
norm = mpl.colors.Normalize(vmin=-2, vmax=2)
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

plt.rcParams['figure.dpi'] = 300
plt.rcParams['figure.figsize'] = 1.07, 2.2
fig, ax = plt.subplots()

# ax.axis('off')

cbar = plt.colorbar(
    sm,
    cax=ax,
    orientation='vertical',
    shrink=0.5,
)

cbar.set_ticks([-2, 2])
cbar.set_ticklabels(['< -5', '> 5'])
cbar.ax.tick_params(labelsize=11)
cbar.set_label('log$_2$ fold change', rotation=90, labelpad=0, fontsize=12)

plt.tight_layout()
plt.show()
```
