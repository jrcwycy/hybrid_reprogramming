```python
print(f"Initial shape: {adata_raw.shape}")
init_cells_hybrid = adata_raw[adata_raw.obs['dataset'] == 'Hybrid'].n_obs

# Identify Hybrid cells without assigned condition
hybrid_to_remove = (adata_raw.obs['dataset'] == 'Hybrid') & (adata_raw.obs['assigned_condition'].isna())

# Create a boolean mask for all cells to keep (non-Hybrid OR Hybrid with assigned condition)
cells_to_keep = ~(hybrid_to_remove)

# Filter the AnnData object
adata_raw = adata_raw[cells_to_keep].copy()

final_cells_hybrid = adata_raw[adata_raw.obs['dataset'] == 'Hybrid'].n_obs
removed_hybrid_cells = init_cells_hybrid - final_cells_hybrid

print(f"Removed {removed_hybrid_cells} 'Hybrid' cells with unassigned feature barcodes")
print(f"Final shape: {adata_raw.shape}")
```

```python
def fastq_to_df(file_path):
    """Function to read in a fastq file and create a dataframe with read IDs and reads"""
    
    seq_ids = []
    reads = []
    q_scores = []
    
    with open(file_path, "r") as handle: #with open(file_path, "r"); gzip.open(file_path, "rt")
        for record in SeqIO.parse(handle, "fastq"):
            seq_ids.append(record.id)
            reads.append(str(record.seq))
            q_scores.append(np.mean(record.letter_annotations["phred_quality"]))
            
    df = pd.DataFrame({'read_id' : seq_ids, 'read' : reads, 'avg_qual' : q_scores})
    return df




df = fastq_to_df(fpath)
df['length'] = df['read'].apply(len)

print(f"{df.shape=}")
df.head()
```

```python
def avg_phred_quality(qual_str):
    
    scores = [ord(c) - 33 for c in qual_str]
    avg_score = sum(scores) / len(scores)
    
    return avg_score
```

```python
df['avg_quality_umi'] = df['quality_umi'].apply(avg_phred_quality)

df['avg_quality_bc'] = df['quality_barcode'].apply(avg_phred_quality)

df.head()
```
